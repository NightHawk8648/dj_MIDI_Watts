## Walkthrough: Android Network Security & VPN Safeguards
**Completed: June 27, 2026**

### 1. Network Security Scoped Configuration
Because the local Windows host is locked down to `LocalSubnet` via Windows Firewall, the Android App's cleartext security needed tightening.
- **`AndroidManifest.xml`:** Removed the insecure, global `android:usesCleartextTraffic="true"`.
- **`network_security_config.xml` Created:** Defined an explicit domain policy that **only** permits cleartext HTTP connections to local subnets (`192.168.x.x`, `10.x.x.x`, `172.16.x.x`, `127.0.0.1`). Any attempt to communicate with an external public IP now strictly enforces HTTPS.

### 2. Network Monitoring & UI Warnings
The strict `LocalSubnet` firewall means the Android app will completely fail to connect if it switches to **Mobile Data** or uses a **Commercial VPN** (as it routes traffic through a different subnet).
- **`NetworkMonitor.kt` Created:** A new background service utilizing Android's `ConnectivityManager`. It actively listens for `TRANSPORT_CELLULAR` (Mobile Data) and `TRANSPORT_VPN`.
- **Permissions Added:** Injected `POST_NOTIFICATIONS` and `FOREGROUND_SERVICE` into the Manifest.
- **Behavior:** If the user drops off the local Wi-Fi onto Mobile Data, or spins up a VPN, the `NetworkMonitor` triggers an immediate system Push Notification alerting the user that DJ Grid connectivity is lost and they must switch back to the local Wi-Fi to regain control.
